```python

import random

# Execution time of each task on each VM
execution_time = [
    [2, 5, 4],
    [6, 2, 3],
    [4, 3, 1],
    [3, 4, 2],
    [5, 2, 4],
    [2, 3, 5],
    [4, 1, 3],
    [3, 5, 2],
    [6, 3, 2],
    [2, 4, 3]
]

num_tasks = len(execution_time)
num_vms = 3
num_ants = 10
iterations = 20

# Initially, all VMs have equal pheromone
pheromone = [[1.0] * num_vms for _ in range(num_tasks)]

alpha = 1.0
beta = 2.0
rho = 0.3

best_assignment = None
best_time = float("inf")

for iteration in range(iterations):
    for ant in range(num_ants):
        assignment = []
        loads = [0] * num_vms

        for task in range(num_tasks):
            # Prefer faster VMs and higher pheromone
            weights = [
                pheromone[task][vm] ** alpha *
                (1 / execution_time[task][vm]) ** beta
                for vm in range(num_vms)
            ]

            total = sum(weights)
            probabilities = [w / total for w in weights]

            vm = random.choices(
                range(num_vms), weights=probabilities
            )[0]

            assignment.append(vm)
            loads[vm] += execution_time[task][vm]

        # Objective: minimize the largest VM workload
        total_time = max(loads)

        if total_time < best_time:
            best_time = total_time
            best_assignment = assignment[:]

        # Evaporate pheromone
        for task in range(num_tasks):
            for vm in range(num_vms):
                pheromone[task][vm] *= (1 - rho)

        # Reward assignments from this ant
        reward = 1 / total_time
        for task, vm in enumerate(assignment):
            pheromone[task][vm] += reward

print("Best task assignment:")

for task, vm in enumerate(best_assignment):
    print(f"Task {task + 1} -> VM {vm + 1}")

print("Estimated total completion time:", best_time)
```
