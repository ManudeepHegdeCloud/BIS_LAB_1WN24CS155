```python

import statistics

image = [
    [50, 52, 49, 51],
    [48, 250, 53, 50],
    [51, 49, 5, 52],
    [50, 54, 51, 49]
]

rows = 4
cols = 4
threshold = 30

result = [row[:] for row in image]

for i in range(rows):
    for j in range(cols):

        neighbors = []

        for x in range(max(0, i-1), min(rows, i+2)):
            for y in range(max(0, j-1), min(cols, j+2)):

                if x != i or y != j:
                    neighbors.append(image[x][y])

        median = statistics.median(neighbors)

        if abs(image[i][j] - median) > threshold:

            normal_neighbors = [
                v for v in neighbors
                if abs(v - median) <= threshold
            ]

            result[i][j] = round(
                sum(normal_neighbors) / len(normal_neighbors)
            )

print("Original Image:")
for row in image:
    print(row)

print("\nAfter Noise Reduction:")
for row in result:
    print(row)

print("\nNoise Reduction Completed!")
```
